```python
import pandas as pd
import unicodedata
from google.colab import files

# Pujo l'arxiu
uploaded = files.upload()
empleats = list(uploaded.keys())[0]

# Carreguem el dataset i ens saltem les 3 primeres files, indicant que el separador és el punt i coma.
df = pd.read_csv(
    empleats,
    skiprows=3,
    sep=",",
    header=0
)

def normalitzar_columnes(df):

    noves = []

    for c in df.columns:

        c = str(c).strip().lower()

        c = unicodedata.normalize(
            "NFKD",
            c
        ).encode(
            "ascii",
            "ignore"
        ).decode()

        c = c.replace(" ", "_")

        noves.append(c)

    df.columns = noves

    return df

df = normalitzar_columnes(df)

print(df.columns)



print(df["salari_mensual"].dtype)

print(df["salari_mensual"].describe())

# Dimensions del dataset
print("Dimensions del dataset:")
print(df.shape)

# Files i columnes
print(f"\nNombre de files: {df.shape[0]}")
print(f"Nombre de columnes: {df.shape[1]}")

# Primeres files
print("\nPrimeres 5 files:")
display(df.head())

# Últimes files
print("\nÚltimes 5 files:")
display(df.tail())

# Informació general
print("\nInformació del dataset:")
df.info()

# Tipus de dades
print("\nTipus de dades:")
display(df.dtypes)

# Estadístiques descriptives
print("\nEstadístiques de variables numèriques:")
display(df.describe())

# Estadístiques variables categòriques
print("\nVariables categòriques:")
display(df.describe(include="object"))

# Valors nuls
print("\nValors nuls per columna:")
display(df.isnull().sum())

# Files duplicades
duplicats = df.duplicated().sum()

print(f"\nNombre de files duplicades: {duplicats}")
```

## <font color = "pink"> Exercici 2: Validacions bàsiques de les dades i decisions d'eliminació </font>
🎯 Objectiu: assegurar que les dades compleixen unes condicions mínimes de qualitat i determinar quines files poden continuar sent analitzades i quines s'han de descartar.

En aquest exercici hauràs de validar aspectes clau del dataset que són crítics en qualsevol context real d'anàlisi de dades, prenent decisions justificades sobre la validesa o eliminació de registres.


Aspectes a validar:


» Existeixen identificadors únics (DNI) per a cada persona? <font color = "yellow"> Sí, tots els registres tenen el seu DNI </font>

» Què faràs amb les files duplicades o registres repetits? <font color = "yellow"> Si estàn exatament duplicades en tots els seus camps, eliminar-ne una. Si està duplicada la PK(DNI) però la resta no concorda, eliminar tots els registres amnb la mateixa PK, ja que es impossible discernir quin d'ells es el correcte </font>

» Hi ha valors obligatoris que falten o informació incoherent? <font color ="yellow">Si. Hi ha camps buits </font>

» Detectes formats incorrectes, errors de codificació o valors impossibles?

Concretament, hauràs de:

» Corregir possibles problemes de codificació i accentuació en els noms de les columnes i en els valors de text, si escau.

» Detectar i eliminar files duplicades exactes, indicant quants registres s'han eliminat.

» Comprovar la unicitat del DNI. En cas de detectar DNIs duplicats, hauràs d'analitzar la coherència de la informació associada i descartar o anul·lar els registres que no siguin fiables, justificant clarament la decisió presa.

» Identificar els camps clau que no haurien de contenir valors nuls, detectar registres amb informació insuficient i eliminar únicament aquelles files completament buides o amb informació irrecuperable o no substituïble, explicant per què no és possible ni raonable imputar-les.

 Important


en aquest exercici no s'espera que imputis dades, sinó que decideixis quins registres poden formar part de l'anàlisi i quins no, argumentant sempre el criteri utilitzat.


🧠 Si el text "s'entén però està mal escrit" → codificació


Si el text "està ben escrit però no és útil per a l'anàlisi" → validació

```python

import pandas as pd


# 1. ELIMINAR DUPLICATS EXACTES

files_abans = len(df)

df = df.drop_duplicates()

files_despres = len(df)

print(f"Duplicats exactes eliminats: {files_abans - files_despres}")

# 2. COMPROVAR DNI DUPLICATS

dni_duplicats = df[df.duplicated(subset="dni", keep=False)]

print(f"\nNombre de registres amb DNI duplicat: {len(dni_duplicats)}")

if len(dni_duplicats) > 0:

    display(
        dni_duplicats.sort_values("dni")
    )

    # Comptem quantes dades té cada registre
    df["completitud"] = df.notna().sum(axis=1)

    # Conservem el registre amb més informació
    df = (
        df
        .sort_values(
            by=["dni", "completitud"],
            ascending=[True, False]
        )
        .drop_duplicates(
            subset="dni",
            keep="first"
        )
    )

    # Eliminem la columna auxiliar
    df.drop(columns="completitud", inplace=True)

print(f"\nNombre de registres després de revisar els DNI: {len(df)}")

# 3. DETECTAR VALORS NULS

print("\nValors nuls per columna:")

display(
    df.isnull()
      .sum()
      .sort_values(ascending=False)
)

# 4. ELIMINAR REGISTRES IRRECUPERABLES

camps_obligatoris = [
    "dni",
    "nom",
    "cognoms",
    "genere"
]

registres_invalids = df[
    df[camps_obligatoris]
      .isnull()
      .any(axis=1)
]

print("\nRegistres eliminats per manca d'informació obligatòria:")

display(registres_invalids)

df = df.dropna(
    subset=camps_obligatoris,
    how="any"
)

print(f"\nNombre de registres després d'eliminar els registres invàlids: {len(df)}")

# 5. FILES COMPLETAMENT BUIDES


files_buides = df.isnull().all(axis=1).sum()

print(f"\nFiles completament buides: {files_buides}")

# 6. RESUM FINAL

print("=" * 50)
print("RESUM DE LA VALIDACIÓ")
print("=" * 50)

print(f"Nombre final de registres: {len(df)}")
print(f"DNI únics: {df['dni'].nunique()}")

print("\nValors nuls restants:")

display(df.isnull().sum())
```

## <font color = "pink"> Exercici 3: Transformacions necessàries i preparació del dataset </font>
🎯 Objectiu: preparar el dataset perquè sigui coherent, analitzable i estructuralment correcte abans de realitzar qualsevol anàlisi descriptiva.

En aquest exercici hauràs d'aplicar transformacions bàsiques però essencials, orientades a millorar la qualitat de les dades i facilitar l'anàlisi posterior.


Aspectes a revisar i transformar:


» Els valors categòrics estan normalitzats i són consistents?

» Els valors nuls són coherents amb la informació disponible?

» Hi ha redundància de variables que pugui generar problemes analítics?

» Les variables contextuals (com temperatura, salari) estan completes i preparades per a l'anàlisi?

Concretament, hauràs de:


» Normalitzar els valors categòrics per evitar ambigüitats (per exemple, valors que representen el mateix concepte però estan escrits de manera diferent).

» Revisar la coherència dels valors nuls, assegurant que:
Les persones que tenen quilometratge anual i consum mitjà associats disposen de cotxe. En els casos en què consti consum mitjà però no quilometratge anual, caldrà decidir com resoldre aquesta incoherència de manera justificada.

Les persones que tenen un valor en el nombre de fills efectivament tenen fills.

» Unificar la informació sobre fills en una única columna, ja que mantenir diverses columnes amb informació complementària no és òptim i pot generar problemes analítics (redundància o multicolinealitat).

» Validar la temperatura mitjana. En cas que aquesta informació falti, hauràs d'imputar-la utilitzant un criteri raonable (per ciutat o per un nivell superior justificat).

» Revisar i imputar els salaris faltants, aplicant criteris analítics coherents (com ara càrrec, grup professional, ciutat, gènere o franja d'edat).

 Important


en aquest exercici no es tracta d'eliminar registres, sinó de transformar, normalitzar i preparar les dades perquè siguin analitzables de manera rigorosa en els nivells següents.

```python
# NORMALITZACIÓ DE VARIABLES CATEGÒRIQUES

columnes_text = df.select_dtypes(include="object").columns

for columna in columnes_text:
    df[columna] = (
        df[columna]
        .astype(str)
        .str.strip()
        .str.lower()
    )

# NORMALITZACIÓ DEL GÈNERE

df["genere"] = df["genere"].replace({

    "h": "home",
    "m": "home",
    "home": "home",

    "d": "dona",
    "f": "dona",
    "dona": "dona",

    "??": "desconegut",
    "nc": "desconegut",
    "a": "desconegut"

})

# NORMALITZACIÓ DE TE_COTXE

df["te_cotxe"] = (
    df["te_cotxe"]
    .replace({
        True: "si",
        False: "no",
        "true": "si",
        "false": "no",
        "sí": "si",
        "s": "si",
        "n": "no"
    })
)

print("Valors del gènere:")
display(df["genere"].value_counts())

# COHERÈNCIA DEL COTXE

# Si no té cotxe, no pot tenir km ni consum.

mask = df["te_cotxe"] == "no"

df.loc[mask, "km_anuals"] = pd.NA
df.loc[mask, "consum_mitja_l_100km"] = pd.NA

# Si hi ha consum però no quilometratge,
# imputem la mediana dels km.

mask = (
    df["consum_mitja_l_100km"].notna()
) & (
    df["km_anuals"].isna()
)

df.loc[mask, "km_anuals"] = df["km_anuals"].median()

# INFORMACIÓ DELS FILLS

df["nombre_fills"] = df["nombre_fills"].fillna(0)

df.loc[
    df["nombre_fills"] < 0,
    "nombre_fills"
] = 0

df["nombre_fills"] = df["nombre_fills"].astype(int)

# TEMPERATURA

df["temperatura_mitjana_ciutat"] = (
    df.groupby("ciutat")["temperatura_mitjana_ciutat"]
      .transform(lambda x: x.fillna(x.mean()))
)

df["temperatura_mitjana_ciutat"] = (
    df["temperatura_mitjana_ciutat"]
      .fillna(df["temperatura_mitjana_ciutat"].mean())
)

# CONVERTIR EL SALARI A NUMÈRIC

df["salari_mensual"] = (
    df["salari_mensual"]
    .astype(str)
    .str.replace("â‚¬", "", regex=False)
    .str.replace("€", "", regex=False)
    .str.replace("euros", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["salari_mensual"] = pd.to_numeric(
    df["salari_mensual"],
    errors="coerce"
)

# IMPUTACIÓ DELS SALARIS

# Primer pel càrrec

df["salari_mensual"] = (
    df["salari_mensual"]
    .fillna(
        df.groupby("ca_rrec")["salari_mensual"]
          .transform("median")
    )
)

# Després pel grup professional

df["salari_mensual"] = (
    df["salari_mensual"]
    .fillna(
        df.groupby("grup_professional")["salari_mensual"]
          .transform("median")
    )
)

# Finalment amb la mediana global

df["salari_mensual"] = (
    df["salari_mensual"]
    .fillna(
        df["salari_mensual"].median()
    )
)

# COMPROVACIÓ FINAL

print("=" * 50)
print("VALORS NULS RESTANTS")
display(df.isnull().sum())

print("\nTIPUS DE DADES")
display(df.dtypes)

print("\nRESUM DEL SALARI")
display(df["salari_mensual"].describe())
```

## <font color = "pink">Exercici 4: Primeres anàlisis descriptives </font>
🎯 Objectiu: obtenir una primera visió global del dataset un cop s'han aplicat les transformacions bàsiques, mitjançant estadístiques descriptives i comparacions simples entre grups rellevants.

Aspectes a revisar i analitzar:


» Construir taules resum descriptives.

» Fer comparacions bàsiques entre grups rellevants.

Concretament, hauràs de:


» Construir taules resum descriptives de salaris com una taula resum que permeti veure el sou mig, medià, mínim i màxim per Gènere. Ordena la taula en funció del sou mig.

» Crear una taula resum amb el salari mig per gènere (files) i país d'origen (columnes). Afegeix-hi les mitjanes als marges de la taula. Aplica format condicional a la taula per veure en un color més intens els valors més elevats.

```python
# TAULA RESUM DELS SALARIS PER GÈNERE

resum_salaris = (
    df.groupby("genere")["salari_mensual"]
      .agg(
          Salari_Mitja="mean",
          Mediana="median",
          Minim="min",
          Maxim="max"
      )
      .sort_values("Salari_Mitja", ascending=False)
      .round(2)
)

print("Resum de salaris per gènere:")

display(resum_salaris)

# TAULA DINÀMICA

taula = pd.pivot_table(
    df,
    values="salari_mensual",
    index="genere",
    columns="pais_d'origen",
    aggfunc="mean",
    margins=True,
    margins_name="Mitjana"
)

print("\nSalari mitjà per gènere i país d'origen:")

display(taula.round(2))

# FORMAT CONDICIONAL

display(
    taula.style
         .background_gradient(cmap="Purples")
         .format("{:.2f}")
)

# EXTRA - PERCENTATGE DE PERSONES PER GÈNERE

percentatges = (
    df["genere"]
      .value_counts(normalize=True)
      .mul(100)
      .round(2)
      .rename("Percentatge")
)

print("\nPercentatge de persones per gènere:")

display(percentatges)

# EXTRA - RÀTIO SALARIAL

mitjana_general = df["salari_mensual"].mean()

ratio = (
    df.groupby("genere")["salari_mensual"]
      .mean()
      .div(mitjana_general)
      .round(2)
      .rename("Ràtio")
)

print("\nRàtio salarial respecte la mitjana general:")

display(ratio)
```
