```python
inputs_test = [
    "Laura Serra", "608112233", 6, 3, 3, 3,
    "Clara Gómez", "655221199", 3, 3, 3, 2,
    "Sergio López", "644112211", 3, 2, 2, 2,
    "Candidat Error", "611000000", 5, 7, 3, 2,  # Invàlid
    "Marta Ruiz", "699887766", 2, 3, 2, 1,
    "Marc Vidal", "612443322", 6, 3, 3, 3,
    "Júlia Romero", "677998877", 5, 3, 3, 2,
    "Pol Navarro", "699554433", 4, 3, 2, 2,
    "David Ferrer", "677665544", 2, 1, 1, 1,
    "Anna Paredes", "633221144", 1, 2, 1, 0,
    "Anibal Paredes", "644272727", 1, 2, 1, 0,
    "Núria Casals", "600112200", 8, 2, 3, 3
]


def obtenir_dades(dades, index):
    nom = dades[index]
    telefon = dades[index + 1]
    experiencia = dades[index + 2]
    catala = dades[index + 3]
    excel = dades[index + 4]
    titulacio = dades[index + 5]

    return nom, telefon, experiencia, catala, excel, titulacio

print(obtenir_dades(inputs_test, 0))


def validar_dades(experiencia, catala, excel, titulacio):

    if experiencia < 0:
        return False

    if catala < 0 or catala > 3:
        return False

    if excel < 0 or excel > 3:
        return False

    if titulacio < 0 or titulacio > 3:
        return False

    return True

print(validar_dades(5, 3, 3, 2))
print(validar_dades(-1, 3, 3, 2))
print(validar_dades(5, 5, 3, 2))
print(validar_dades(5, 3, 5, 2))
print(validar_dades(5, 3, 2, 6))

def calcular_puntuacio(experiencia, catala, excel, titulacio):

    puntuacio = (experiencia * 1.5 + catala + excel + titulacio)

    return puntuacio

print (calcular_puntuacio(5, 3, 3, 2))


def classificar_candidat(puntuacio):

    if puntuacio >= 12:
        return "Prioritat Alta"

    elif puntuacio >= 8:
        return "Candidat adequat"

    else:
        return "No prioritari"

print(classificar_candidat((15.5)))


def mostrar_individual(nom, puntuacio, categoria):

    print(
        f"Candidat: {nom} | "
        f"Puntuació: {puntuacio:.1f} | "
        f"Categoria: {categoria}"
    )


def mostrar_informe(candidats):

    ordre_prioritat = {
        "Prioritat Alta": 1,
        "Candidat adequat": 2,
        "No prioritari": 3
    }

    candidats_ordenats = sorted(
        candidats,
        key=lambda x: (
            ordre_prioritat[x["categoria"]],
            -x["puntuacio"]
        )
    )

    print("\n\n===== INFORME FINAL =====")

    for candidat in candidats_ordenats:

        print(
            f"Nom: {candidat['nom']} | "
            f"Telèfon: {candidat['telefon']} | "
            f"Puntuació: {candidat['puntuacio']:.1f} | "
            f"Categoria: {candidat['categoria']}"
        )


def main_seleccio(inputs_test):

    candidats = []

    index = 0

    while len(candidats) < 10:

        nom, telefon, experiencia, catala, excel, titulacio = obtenir_dades(
            inputs_test,
            index
        )

        index += 6

        if not validar_dades(
            experiencia,
            catala,
            excel,
            titulacio
        ):

            print(f"Error en les dades de: {nom}")
            continue

        puntuacio = calcular_puntuacio(
            experiencia,
            catala,
            excel,
            titulacio
        )

        categoria = classificar_candidat(puntuacio)

        mostrar_individual(
            nom,
            puntuacio,
            categoria
        )

        candidat = {
            "nom": nom,
            "telefon": telefon,
            "puntuacio": puntuacio,
            "categoria": categoria
        }

        candidats.append(candidat)

    mostrar_informe(candidats)


main_seleccio(inputs_test)
```

```python
import pandas as pd

clients = [
    {"id_client": 1, "nom": "Laura Serra", "edat": 23, "grup_edat": "18-25", "codi_comportament": 4},
    {"id_client": 2, "nom": "Marc Vidal", "edat": 35, "grup_edat": "26-40", "codi_comportament": 3},
    {"id_client": 3, "nom": "Ana López", "edat": 29, "grup_edat": "26-40", "codi_comportament": 2},
    {"id_client": 4, "nom": "Joan Riera", "edat": 19, "grup_edat": "18-25", "codi_comportament": 2},
    {"id_client": 5, "nom": "Mónica Pérez", "edat": 47, "grup_edat": "41-60", "codi_comportament": 4},
    {"id_client": 6, "nom": "Luis García", "edat": 52, "grup_edat": "41-60", "codi_comportament": 1},
    {"id_client": 7, "nom": "Pilar Sánchez", "edat": 61, "grup_edat": "60+", "codi_comportament": 3},
    {"id_client": 8, "nom": "Eva Martín", "edat": 38, "grup_edat": "26-40", "codi_comportament": 0},
    {"id_client": 9, "nom": "Diego Romero", "edat": 24, "grup_edat": "18-25", "codi_comportament": 1},
    {"id_client": 10, "nom": "Núria Costa", "edat": 33, "grup_edat": "26-40", "codi_comportament": 4}
]

def obtenir_segment_i_factor(codi_comportament):

    segment_factor = {
        0: {"segment": "Navega però no afegeix productes",
            "factor_codi": 0.10},
        1: {"segment": "Afegeix productes però no arriba al checkout",
            "factor_codi": 0.25},
        2: {"segment": "Arriba al checkout però no paga",
            "factor_codi": 0.50},
        3: {"segment": "Cistella abandonada fa menys de 24 h",
            "factor_codi": 0.75},
        4: {"segment": "Cistella abandonada fa més de 24 h",
            "factor_codi": 0.90}
    }

    return segment_factor.get(
        codi_comportament,
        {"segment": "Desconegut",
            "factor_codi": 0}
    )


def obtenir_factor_edat(grup_edat):

    factors = {
        "18-25": 1.20,
        "26-40": 1.00,
        "41-60": 0.85,
        "60+": 0.70
    }

    return factors.get(grup_edat, 1)


def calcular_risc(factor_codi, factor_edat):

    return round(factor_codi * factor_edat, 2)


def processar_clients(clients):

    resultats = []

    for client in clients:

        dades_segment = obtenir_segment_i_factor(client["codi_comportament"])

        factor_edat = obtenir_factor_edat(["grup_edat"])

        risc_final = calcular_risc(
            dades_segment["factor_codi"],
            factor_edat
        )

        resultats.append({
            "id_client": client["id_client"],
            "nom": client["nom"],
            "edat": client["edat"],
            "grup_edat": client["grup_edat"],
            "codi_comportament": client["codi_comportament"],
            "segment": dades_segment["segment"],
            "factor_codi": dades_segment["factor_codi"],
            "factor_edat": factor_edat,
            "risc_final": risc_final
        })

    return resultats


def mostrar_dataframe(resultats):

    df = pd.DataFrame(resultats)

    df = df.sort_values(
        by="risc_final",
        ascending=False
    )

    print(df)

    return df


def mostrar_respostes(df):

    print("\nCLIENTS PRIORITARIS")
    print(df[["nom", "risc_final"]].head(3))

    print("\nCLIENTS AMB BAIX INTERÈS")
    print(
        df[df["risc_final"] < 0.20][
            ["nom", "risc_final"]
        ]
    )

    millor_client = df.iloc[0]

    print("\nCLIENT MÉS RENDIBLE DE RECUPERAR")

    print(
        f"{millor_client['nom']} "
        f"({millor_client['segment']}) "
        f"- risc: {millor_client['risc_final']}"
    )


def main_marketing(clients):

    resultats = processar_clients(clients)

    df = mostrar_dataframe(resultats)

    mostrar_respostes(df)


main_marketing(clients)
```

```python
import pandas as pd

pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}

def normalitzar_simptomes(string):
  simptomes = string.lower().split(",")

  resultat = []

  for simptoma in simptomes:
    simptoma = simptoma.strip()
    simptoma = simptoma.replace(" ", "_")
    resultat.append(simptoma)

  return resultat


def calcular_risc(simptomes):
    risc = 0

    for simptoma in simptomes:
        risc += nivells_risc.get(simptoma, 0)

    return risc

def processar_pacients():
    resultat = []

    for pacient in pacients:

        nom, edat, simptomes = pacient.split("|")

        simptomes = normalitzar_simptomes(simptomes)

        resultat.append({
            "nom": nom,
            "edat": int(edat),
            "simptomes": simptomes,
            "risc_total": calcular_risc(simptomes)
        })

    return resultat


dades = processar_pacients()

df = pd.DataFrame(dades)

df = df.sort_values("risc_total", ascending=False)

print(df)

```

```python
import pandas as pd

clients = [
    {"nom": "Ana", "edat": 42, "codi": 0},
    {"nom": "Carlos", "edat": 30, "codi": 2},
    {"nom": "Isabel", "edat": 55, "codi": 3},
    {"nom": "Jorge", "edat": 40, "codi": 1},
    {"nom": "Marta", "edat": 28, "codi": 0}
]

nivells_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}


def calcular_preu_final(preu_base, edat):
    return preu_base * (1 + edat / 100)


def processar_clients():

    resultat = []

    for client in clients:

        info = nivells_seguro[client["codi"]]

        resultat.append({
            "Nom": client["nom"],
            "Edat": client["edat"],
            "Categoria": info["categoria"],
            "Preu_base": info["preu_base"],
            "Preu_final": calcular_preu_final(
                info["preu_base"],
                client["edat"]
            ),
            "Probabilitat_frau": info["fraude"]
        })

    return resultat


dades = processar_clients()

df = pd.DataFrame(dades)

print(df)

print("\nOrdenat per frau")
print(df.sort_values("Probabilitat_frau", ascending=False))

print("\nOrdenat per preu")
print(df.sort_values("Preu_final", ascending=False))

print("\nMés frau:")
print(df.loc[df["Probabilitat_frau"].idxmax()])

print("\nPreu més alt:")
print(df.loc[df["Preu_final"].idxmax()])
```

## Exercici 1: Generador de contrasenyes segures
Ets analista de dades dins el departament de Ciberseguretat d'una empresa tecnològica. S'ha detectat que la majoria de treballadors continuen utilitzant contrasenyes molt dèbils com “Asdf1234”, dates d'aniversari o combinacions fàcils d'endevinar. Per aquest motiu, el responsable del departament t'ha demanat desenvolupar un generador de contrasenyes segures, totalment parametritzable i modular.

📝 Requisits del generador

Has de crear una funció principal i totes les subfuncions necessàries per:

» Generar contrasenyes de longitud variable.

» Decidir si han d'incloure:

Majúscules

Minúscules

Números

Signes especials

» Assegurar-te que com a mínim apareix un caràcter de cada tipus seleccionat.

» Aleatoritzar completament la contrasenya.

» Retornar la contrasenya generada.

📍 Nota important
Explora el funcionament del mòdul random de la llibreria icona enllaç numpy

La funció principal del generador ha de tenir valors per defecte, exactament els següents:


majuscules=True

minuscules=True

numeros=True

signes=False

Aquests paràmetres són obligatoris i formen part del disseny del sistema.

```python
import numpy as np
import string

def obtenir_caracters(majuscules, minuscules, numeros, signes):
    conjunts = []
    pool_total = ""

    if majuscules:
        conjunts.append(string.ascii_uppercase)
        pool_total += string.ascii_uppercase
    if minuscules:
        conjunts.append(string.ascii_lowercase)
        pool_total += string.ascii_lowercase
    if numeros:
        conjunts.append(string.digits)
        pool_total += string.digits
    if signes:
        conjunts.append(string.punctuation)
        pool_total += string.punctuation

    return conjunts, pool_total

def generar_contrasenya(longitud=12, majuscules=True, minuscules=True, numeros=True, signes=False):
    conjunts_obligatoris, pool_total = obtenir_caracters(majuscules, minuscules, numeros, signes)

    if longitud < len(conjunts_obligatoris):
        raise ValueError("La longitud ha de ser igual o major al nombre de tipus de caràcters seleccionats.")

    if not pool_total:
        raise ValueError("Has de seleccionar com a mínim un tipus de caràcter.")

    contrasenya = []


    for conjunt in conjunts_obligatoris:
        contrasenya.append(np.random.choice(list(conjunt)))

    espai_restant = longitud - len(contrasenya)
    if espai_restant > 0:
        contrasenya.extend(np.random.choice(list(pool_total), size=espai_restant))

    np.random.shuffle(contrasenya)

    return "".join(contrasenya)

pwd_generada = generar_contrasenya(longitud=16, majuscules=True, minuscules=True, numeros=True, signes=True)
print(f"Contrasenya generada: {pwd_generada}")
```

## Exercici 2: Processament automàtic de dades esportives
Ara formes part d'un equip d'analítica dins una organització esportiva que gestiona dades de la lliga catalana.
```python
import pandas as pd
from google.colab import files


def inicialitzar_equip(equips: dict, nom: str) -> None:
    """Afegeix un equip al diccionari si encara no hi és,
    inicialitzant totes les estadístiques a zero."""
    if nom not in equips:
        equips[nom] = {
            "gols_fets":   0,
            "gols_rebuts": 0,
            "victòries":   0,
            "empats":      0,
            "derrotes":    0,
            "punts":       0,
        }


def processar_linia(equips: dict, linia: str) -> None:
    """Llegeix una línia amb format 'local [TAB] gf-gc [TAB] visitant'
    i actualitza les estadístiques dels dos equips."""
    local, resultat, visitant = linia.strip().split("\t")
    g_local, g_visitant = resultat.split("-")
    g_local    = int(g_local)
    g_visitant = int(g_visitant)

    inicialitzar_equip(equips, local)
    inicialitzar_equip(equips, visitant)

    equips[local]["gols_fets"]      += g_local
    equips[local]["gols_rebuts"]    += g_visitant
    equips[visitant]["gols_fets"]   += g_visitant
    equips[visitant]["gols_rebuts"] += g_local

    if g_local > g_visitant:
        equips[local]["punts"]       += 3
        equips[local]["victòries"]   += 1
        equips[visitant]["derrotes"] += 1
    elif g_local < g_visitant:
        equips[visitant]["punts"]     += 3
        equips[visitant]["victòries"] += 1
        equips[local]["derrotes"]     += 1
    else:
        equips[local]["punts"]     += 1
        equips[local]["empats"]    += 1
        equips[visitant]["punts"]  += 1
        equips[visitant]["empats"] += 1


def llegir_fitxer(ruta: str) -> dict:
    """Obre el fitxer de partits i retorna el diccionari d'equips."""
    equips = {}
    with open(ruta, "r", encoding="utf-8") as fitxer:
        for linia in fitxer:
            if linia.strip():
                processar_linia(equips, linia)
    return equips


def construir_dataframe(equips: dict) -> pd.DataFrame:
    """Converteix el diccionari en un DataFrame ordenat per punts."""
    df = pd.DataFrame.from_dict(equips, orient="index")
    df.index.name = "Equip"
    df = df.rename(columns={
        "gols_fets":   "GF",
        "gols_rebuts": "GC",
        "victòries":   "V",
        "empats":      "E",
        "derrotes":    "D",
        "punts":       "Pts",
    })
    df = df[["V", "E", "D", "GF", "GC", "Pts"]]
    df = df.sort_values("Pts", ascending=False).reset_index()
    df.index += 1                           # posició des d'1
    return df

def equip_mes_golejador(equips: dict) -> str:
    """Retorna el nom de l'equip amb més gols marcats."""
    return max(equips, key=lambda e: equips[e]["gols_fets"])


def equip_mes_golejat(equips: dict) -> str:
    """Retorna el nom de l'equip amb més gols encaixats."""
    return max(equips, key=lambda e: equips[e]["gols_rebuts"])


def imprimir_resultats(equips: dict, df: pd.DataFrame) -> None:
    """Mostra per pantalla el resum complet de la lliga."""
    golejador = equip_mes_golejador(equips)
    golejat   = equip_mes_golejat(equips)

    print("=" * 50)
    print("  RESUM DE LA LLIGA CATALANA")
    print("=" * 50)

    print(f"\nEquip més golejador: {golejador} ({equips[golejador]['gols_fets']} gols marcats)")
    print(f"Equip més golejat:   {golejat} ({equips[golejat]['gols_rebuts']} gols encaixats)")

    print("\nTotal de gols marcats per equip:")
    for equip, stats in sorted(equips.items(),
                                key=lambda x: x[1]["gols_fets"],
                                reverse=True):
        print(f"  {equip}: {stats['gols_fets']} gols")

    print("\nClassificació final:")
    print(df.to_string())

    print("\nResultat esperat:")
    print(f"  Equip més golejador: {golejador}")
    print(f"  Equip més golejat:   {golejat}")
    print("\n  Classificació final:")
    for pos, fila in df.iterrows():
        print(f"  {pos}. {fila['Equip']}  —  {fila['Pts']} pts")


def main():
    """Punt d'entrada principal del programa."""
    uploaded = files.upload()
    ruta = list(uploaded.keys())[0]

    equips = llegir_fitxer(ruta)
    df     = construir_dataframe(equips)
    imprimir_resultats(equips, df)

    return df, equips


if __name__ == "__main__":
    df, equips = main()
```
