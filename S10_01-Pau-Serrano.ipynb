4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

Mostra el codi d'estat.

6. Fes una petició DELETE sobre una publicació.

Mostra el codi d'estat.

```python
import requests


# Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir: Llista de publicacions (/posts) Llista d’usuaris (/users) Llista de tasques (/todos)
posts_response = requests.get("https://jsonplaceholder.typicode.com/posts")
users_response = requests.get("https://jsonplaceholder.typicode.com/users")
todos_response = requests.get("https://jsonplaceholder.typicode.com/todos")

posts_data = posts_response.json()
users_data = users_response.json()
todos_data = todos_response.json()

# Mostra per pantalla: La quantitat total de cada recurs. El codi d'estat de cada petició.
print(f"Posts - Quantitat total: {len(posts_data)}, Codi d'estat: {posts_response.status_code}")
print(f"Users - Quantitat total: {len(users_data)}, Codi d'estat: {users_response.status_code}")
print(f"Todos - Quantitat total: {len(todos_data)}, Codi d'estat: {todos_response.status_code}")
print("----------------------------------")

# Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.
url_inexistent = "https://jsonplaceholder.typicode.com/posts/999999"
resposta_404 = requests.get(url_inexistent)

print("--- PETICIÓ A RECURS INEXISTENT (404) ---")
print(f"Petició feta a: {url_inexistent}")
print(f"Codi d'estat rebut: {resposta_404.status_code}")
print("----------------------------------")


# Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.
# Mostra la resposta JSON.
# Mostra el codi d'estat.

nova_publicacio = {
    "title": "Publicacio fictícia 1",
    "body": "Aquesta publicació pertany a l'exercici 1 de l'Sptinr 10 de la It Academy de Barcelona Activa",
    "userId": 42
}

url_posts = "https://jsonplaceholder.typicode.com/posts"
resposta_post = requests.post(url_posts, json=nova_publicacio)

print("--- PETICIÓ A POST AMB UNA NOVA PUBLICACIÓ ---")
print(f"Codi d'estat: {resposta_post.status_code}")
print("Resposta JSON obtinguda:")
print(resposta_post.json())
print("----------------------------------")

# Fes una petició PATCH per modificar parcialment una publicació existent.
# Mostra la resposta JSON.
# Mostra el codi d'estat.

dades_modificades = {
    "title": "Aquest títol ha estat modificat amb PATCH"
}

url_patch = "https://jsonplaceholder.typicode.com/posts/101"
resposta_patch = requests.patch(url_patch, json=dades_modificades)

print("--- PETICIÓ A PATCH PER MODIFICAR UNA PUBLICACIÓ ---")
print(f"Codi d'estat: {resposta_patch.status_code}")
print("Resposta JSON de la modificació:")
print(resposta_patch.json())
print("----------------------------------")

# Fes una petició DELETE sobre una publicació.
# Mostra la resposta JSON.
# Mostra el codi d'estat.

url_delete = "https://jsonplaceholder.typicode.com/posts/101"
resposta_delete = requests.delete(url_delete)


print("--- PETICIÓ DELETE PER A ELIMINAR UNA PUBLICACIÓ ---")
print(f"Codi d'estat: {resposta_delete.status_code}")
print("Resposta JSON de l'eliminació:")
print(resposta_delete.json())
```
